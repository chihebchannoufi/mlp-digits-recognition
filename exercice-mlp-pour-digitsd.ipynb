```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
import joblib
```

```python
print("Chargement du dataset digits...")
digits = load_digits()
X = digits.data
y = digits.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
print("Transformation (Standardisation) des données manuellement...")
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("Définition du modèle...")
mlp = MLPClassifier(max_iter=1000, random_state=42)

#Recherche d'hyperparamètres directement sur le MLP
param_grid = {
    "hidden_layer_sizes": [(64,), (128,), (64, 32), (128, 64)],
    "activation": ["relu", "tanh"],
    "solver": ["adam", "sgd"],
    "alpha": [1e-4, 1e-2],
    "learning_rate_init": [1e-3, 1e-2],
}
```

```python
print("Lancement de la recherche des meilleurs hyperparamètres (GridSearchCV)...")
grid_search = GridSearchCV(mlp, param_grid, cv=3, n_jobs=-1, verbose=1)
```

```python
# On entraîne le GridSearchCV sur les données déjà transformées (scalées)
grid_search.fit(X_train_scaled, y_train)

print("\n--- RÉSULTATS ---")
print("Meilleurs paramètres trouvés :", grid_search.best_params_)
```

```python
# Évaluation
score_test = grid_search.score(X_test_scaled, y_test)
print(f"Précision du modèle sur les tests : {score_test * 100:.2f}%")
```

```python
# 6. Sauvegarde groupée du scaler et du meilleur modèle
meilleur_modele = grid_search.best_estimator_
paquet_a_sauvegarder = {
    'scaler': scaler,
    'modele': meilleur_modele
}

joblib.dump(paquet_a_sauvegarder, 'modele_et_scaler.pkl')
print("\nModèle et scaler sauvegardés avec succès sous 'modele_et_scaler.pkl' !")
```
